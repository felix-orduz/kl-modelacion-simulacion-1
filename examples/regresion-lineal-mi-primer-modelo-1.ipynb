# Primer Modelo de Predicción con Regresión Lineal (SIMPLE)
 
 
## La regresión lineal es un algoritmo estadístico fundamental que modela la relación lineal entre una variable dependiente y una o más variables independientes. Es la base para muchos métodos de machine learning y proporciona interpretabilidad y simplicidad.

## Modelo de Regresión Lineal Simple Para una variable predictora:
 
# $$y = \beta_0 + \beta_1 x + \epsilon$$

## donde:
## - $y$: Variable dependiente (target)
## - $x$: Variable independiente (predictor)
## - $\beta_0$: Intercepto (término constante)
## - $\beta_1$: Pendiente (coeficiente)
## - $\epsilon$: Error aleatorio

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

# Generación de datos aleatoria

```python
np.random.seed(42)
x = np.random.uniform(0, 10, 30)
y = 2.5 * x + 3 + np.random.normal(0, 2, 30)

# Gráfica datos originales
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, color='blue')
plt.title('Datos Originales')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.show()
```

# Creamos el modelo de predicción lineal

```python
X = x.reshape(-1, 1)
modelo = LinearRegression()
modelo.fit(X, y)

# Gráfica con modelo
x_line = np.linspace(0, 10, 100).reshape(-1, 1)
y_line = modelo.predict(x_line)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, color='blue', label='Datos')
plt.plot(x_line, y_line, 'r-', linewidth=3, 
         label=f'y = {modelo.coef_[0]:.2f}x + {modelo.intercept_:.2f}')
plt.title('Regresión Lineal Ajustada')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

# Creamos un conjunto de predicciones con nuestro modelo

```python
nuevos_x = [[-1], [2], [6], [9], [11]]
predicciones = modelo.predict(nuevos_x)

print("Predicciones:")
for x_val, y_pred in zip(nuevos_x, predicciones):
    print(f"x = {x_val[0]:3.1f} → ŷ = {y_pred:5.2f}")
```

# Graficamos las predicciones y los datos originales

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, color='blue')
plt.plot(x_line, y_line, 'r-', linewidth=3)

# Estas 2 líneas grafican las predicciones:
plt.scatter(nuevos_x, predicciones, color='green', s=100, marker='s')
for x_val, y_pred in zip(nuevos_x, predicciones):
    plt.text(x_val[0], y_pred, f' {y_pred:.1f}', va='bottom')

plt.title('Regresión Lineal con Predicciones')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, color='blue', label='Datos')
plt.plot(x_line, y_line, 'r-', linewidth=3, 
         label=f'y = {modelo.coef_[0]:.2f}x + {modelo.intercept_:.2f}')

# Solo esto para graficar predicciones:
plt.scatter(nuevos_x, predicciones, color='red', s=80, marker='x', 
           linewidth=2, zorder=5, label='Predicciones')

plt.title('Regresión Lineal con Predicciones')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
